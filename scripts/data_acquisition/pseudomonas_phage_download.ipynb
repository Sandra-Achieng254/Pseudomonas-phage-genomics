# Kenyan sequences

## KEN19 strain

```python
from Bio import Entrez, SeqIO

# Always provide your email
Entrez.email = "otienosandra19@gmail.com" # change to siut your email

# List of accession numbers 
accessions = [
    "PP456878"
]

# Output file
output_file = "C:/Users/43676/Downloads/KEN19_sequences.fasta" # change to the desired path on your computer

# Download and write sequences to FASTA
with open(output_file, "w") as out_handle:
    for acc in accessions:
        try:
            handle = Entrez.efetch(db="nucleotide", id=acc, rettype="fasta", retmode="text")
            seq_record = handle.read()
            out_handle.write(seq_record)
            print(f"Downloaded {acc}")
        except Exception as e:
            print(f"Failed to download {acc}: {e}")

# if 1 fails  it will not appear in the final .fasta file, delete it and try again or download it separately if it keeps refusing
```
